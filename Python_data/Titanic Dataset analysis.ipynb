# Data Preprocessing

```python
#check missing values
df.isna().sum()

```

```python
# manage missing values as you see pertinent
age_moy = df["age"].mean()
age_moye = int(age_moy)
df["age"] = df["age"].fillna(age_moye)




```

```python
#check missing values
df.isna().sum()


```

```python
# manage missing values as you see pertinent

df['embarked'] = df['embarked'].fillna('S')
df.isna().sum()
```

```python
# manage missing values as you see pertinent
# manage missing values as you see pertinent
df = df.drop(columns = ['deck'])
df

```

```python
#check missing values
df['embark_town'] = df['embark_town'].fillna('Southampton')

df.isna().sum()
```

```python
# Check for duplicate rows
print(df.duplicated())
print("\nNombre de doublons :", df.duplicated().sum())
```

```python
#Drop duplicates keeping the first occurance
df_final = df.drop_duplicates(keep='first')
print("\nDonnées après suppression des doublons :")
print(df_final)
```

```python
# Check for duplicate rows
print(df_final.duplicated())
print("\nNombre de doublons :", df_final.duplicated().sum())
```
