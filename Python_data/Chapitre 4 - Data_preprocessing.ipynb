```python
# ---------------------------
# 3. Traitement des valeurs manquantes
# ---------------------------

# Identifier les valeurs manquantes
print("\nNombre de valeurs manquantes par colonne :")
print(df.isnull().sum())

```

```python
# Exemple 2 : Remplissage des valeurs manquantes

# Remplir la note manquante par 0 (exemple simple)
df["Note"] = df["Note"].fillna(0)

#OU BIEN
# Remplir l'âge manquant par la moyenne
moyenne_age = df["Age"].mean()
df["Age"] = df["Age"].fillna(moyenne_age)

df
```
